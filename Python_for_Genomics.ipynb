```python
#función para gc%
def gc (dna):
  #n o N son bases indefinidas, por eso se quitan del total después
  nbases = dna.count('n')+dna.count('N')
  gcpercent = (dna.count('c')+dna.count('C')+dna.count('g')+dna.count('G'))/(len(dna)-nbases)*100
  return gcpercent
```

```python
#función para detectar codones de alto
#frame=0 sirve para darle un valor por defecto
def dna_has_stop(dna,frame=0):
  stop_codon_found = False
  codons = ['tga','tag','taa']
  for i in range(frame,len(dna),3):
    codon = dna[i:i+3].lower()
    if codon in codons:
      stop_codon_found = True
      break
  return stop_codon_found
```

```python
def strreverse (seq):
  return seq[::-1]

def complement (dna):
  basecomplement = {'a':'t','c':'g','t':'a','g':'c',
                    'A':'T','C':'G','T':'A','G':'C'}
  letters = list(dna)
  #reemplaza la base complemento según la libreria
  #el for e in recorren la lista
  letters = [basecomplement[base] for base in letters]
  return ''.join(letters)

def reversecomplement(seq):
  seq = strreverse(seq)
  seq = complement(seq)
  return seq
```

```python
def count1(dna, base):
    i = 0
    for c in dna:
        if c == base:
	         i += 1
    return i

def count2(dna, base):
    i = 0
    for j in range(len(dna)):
        if dna[j] == base:
	        i += 1
    return i

def count3(dna, base):
    match = [c == base for c in dna]
    return sum(match)

def count4(dna, base):
    return dna.count(base)

def count5(dna, base):
    return len([i for i in range(len(dna)) if dna[i] == base])

def count6(dna,base):
    return sum(c == base for c in dna)
```

```python
#La idea es ejecutar todas las celdas siguientes sin tener que introducir datos en las funciones siguientes, además de N
#Paso 1: importar archivo desde url

import requests

#exam file = "https://d396qusza40orc.cloudfront.net/genpython/data_sets/dna2.fasta"
#example file = "https://d396qusza40orc.cloudfront.net/genpython/data_sets/dna.example.fasta"


url = input("Ingresa la URL del FASTA: ")
archivo = url.split('/')[-1]

response = requests.get(url)

if response.status_code == 200:
    with open(archivo, 'w') as f:
        f.write(response.text)

    print(f"Archivo '{archivo}' descargado y guardado en Colab.")
else:
    print("Error al descargar el archivo")
```

```python
#Paso 2: dividir por records

#abrir el archivo
try:
  f = open(archivo)
  print (f"ok, archivo abierto")
except IOError:
  print ("This file does not exist")


#leer registros FASTA
seqs = {}
for line in f:
  line = line.rstrip()

  if line[0] == '>':

    words = line.split()
    name = words[0][1:]
    seqs[name] = ""
  else:
    seqs[name] = seqs[name] + line

print (len (seqs))

```

```python
#Paso 3: Analizar longitudes de las secuencias (min/max) e identificadores

def seqlens(seqs):

    longitudes = [len(seq) for seq in seqs.values()]

    minlen = min(longitudes)
    maxlen = max(longitudes)


    print("Longitudes de cada secuencia:")
    for ID, seq in seqs.items():
        print(f"  {ID}:    {len(seq)}")

    print(f"\nLongitud Mínima: {minlen}")
    print(f"Longitud Máxima: {maxlen}")


    #busca el ID de la longitud mínima y máxima
    minID = [ID for ID, seq in seqs.items() if len(seq) == minlen]
    maxID = [ID for ID, seq in seqs.items() if len(seq) == maxlen]


    #  mínima
    print(f"\nSecuencia más corta ({minlen}):")
    if len(minID) > 1:
        print("Hay mas de un minimo " + ", ".join(minID))
    else:
        print(f"ID: {minID[0]}")

    #  máxima
    print(f"\nSecuencia más larga ({maxID}):")
    if len(maxID) > 1:
        print("Hay mas de un máxima ".join(maxID))
    else:
        print(f"Identificador: {maxID[0]}")


seqlens(seqs)
```

```python
#Paso 4: leer ORF´s según el marco N

CODONINIT = "ATG"
CODONALTO = {"TAA", "TAG", "TGA"}

def ORFSenN(sequencia, OFFSET):

    ORFS = []

    INDICE = OFFSET

    SIESORF = False
    INICIAORF = -1


    while INDICE <= len(sequencia) - 3:
        codon = sequencia[INDICE : INDICE + 3]

        if not SIESORF:
            if codon == CODONINIT:
                SIESORF = True
                INICIAORF = INDICE + 1

            INDICE += 3

        else:
            if codon in CODONALTO:
                ORFLEN = (INDICE + 3) - (INICIAORF - 1)

                ORFS.append((ORFLEN, INICIAORF))

                SIESORF = False
                INICIAORF = -1

                INDICE += 3

            else:
                INDICE += 3

    return ORFS






while True:
    try:
        MARCO = int(input("Ingrese el marco de lectura (1, 2 o 3): "))
        if MARCO in [1, 2, 3]:
            OFFSET = MARCO - 1
            break
        else:
            print("Error, solo 1, 2 o 3")
    except ValueError:
        print("Solo numeros")

maxORF = 0
maxORFID = ""
ORFposinit = -1

print(f"\nORFs en el Marco de Lectura {MARCO}")

for SEQid, secuencia in seqs.items():

    orfs = ORFSenN(secuencia, OFFSET)

    maxSeqORF = 0
    maxSeqPosinit = -1

    if orfs:

        TuplaMAXORF = max(orfs)
        maxSeqORF = TuplaMAXORF[0]
        maxSeqPosinit = TuplaMAXORF[1]


        if maxSeqORF > maxORF:
            maxORF = maxSeqORF
            maxORFID = SEQid
            ORFposinit = maxSeqPosinit

    if maxSeqORF > 0:
        print(f"\nSecuencia: {SEQid}")
        print(f" ORF más largo en el Marco {MARCO}: {maxSeqORF}")
        print(f" Posición de inicio: {maxSeqPosinit}")
    else:
         print(f"\nSecuencia: {SEQid}")
         print(f" No se encontraron ORFs completos en el Marco {MARCO}.")


print(f"\n\nTOTALES DEL MARCO {MARCO})")
print(f"\n1. Longitud del ORF más largo en el archivo: {maxORF}")
print(f"\n2. Identificador de la secuencia con el ORF más largo: {maxORFID}")
print(f"\n3. Posición de inicio del ORF más largo (base 1): {ORFposinit}")
```

```python
#Paso 5: leer repeats
from operator import itemgetter

def repeticiones (seqsDict, N):

    reptotales = {}

    for seq_id, seq in seqsDict.items():
        seqslen = len(seq)


        for i in range(seqslen - N + 1):

            substring = seq[i : i + N]

            reptotales[substring] = reptotales.get(substring, 0) + 1

    return reptotales


def analisis(conteoreps, N):
    if conteoreps:
        repmasfreq, maxconteo = max(conteoreps.items(), key=lambda x: x[1])

        maxrepeticiones = [
            rep for rep, count in conteoreps.items()
            if count == maxconteo
        ]

        maxreps = len(maxrepeticiones)
        masmaxima = maxreps > 1

        print(f"\nRepeticiones de Longitud N = {N} ")
        repsunicos = {substr: count for substr, count in conteoreps.items() if count > 1}
        print(f"Número total de repeats unicos encontrados: {len(repsunicos)}")
        print(f"\nLa repetición mas frecuente es: {', '.join(maxrepeticiones)}")
        print(f"Ocurrencias totales en el archivo: {maxconteo} veces")

        if masmaxima:
            print(f"Hay {maxreps} repeticiones con la frecuencia MAX.")
        else:
            print("Solo se encontró una repetición con la frecuencia máxima.")
    else:
        print(f"\nNo se encontraron repeticiones de longitud {N} en las secuencias.")


N = int(input("Ingresa la longitud de las repeticiones: "))
conteo = repeticiones(seqs, N)
analisis(conteo, N)

```
